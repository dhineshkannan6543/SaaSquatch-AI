```python
import urllib.request
import json
import re
from datetime import datetime

# Sample lower-middle market target website URL
TARGET_DOMAIN = 'fieldtracklogistics.com'
TARGET_URL = f'https://{TARGET_DOMAIN}'
print(f'[*] Initializing SaaSquatch AI Pipeline for: {TARGET_URL}')
```

## 1. Web Scraping & DOM Meta Extraction Engine

```python
def scrape_website_metadata(domain):
    # Simulated DOM metadata payload for demonstration
    return {
        'domain': domain,
        'title': 'FieldTrack Logistics Solutions - Field Service Automation',
        'meta_description': 'B2B field service dispatch and scheduling software for HVAC and plumbing contractors.',
        'tech_stack': [
            {'category': 'CMS', 'name': 'WordPress v5.8', 'modernity': 'Legacy'},
            {'category': 'Analytics', 'name': 'Google Analytics UA', 'modernity': 'Outdated'},
            {'category': 'Support', 'name': 'Zendesk Email Ticket', 'modernity': 'Legacy'},
            {'category': 'Framework', 'name': 'PHP / jQuery', 'modernity': 'Legacy'}
        ],
        'has_chat': False,
        'has_blog': True
    }

raw_data = scrape_website_metadata(TARGET_DOMAIN)
print(json.dumps(raw_data, indent=2))
```

## 2. Caprae AI Readiness & Acquisition Scoring Algorithm

```python
def calculate_ai_readiness(tech_stack):
    tech_modernity = 48
    automation_potential = 92 # High headroom for LLM dispatch automation
    support_ai_fit = 88       # Perfect candidate for 24/7 AI Voice Assistant
    data_infra = 55
    
    overall_score = int(tech_modernity * 0.25 + automation_potential * 0.35 + support_ai_fit * 0.25 + data_infra * 0.15)
    
    return {
        'tech_modernity': tech_modernity,
        'automation_potential': automation_potential,
        'support_ai_fit': support_ai_fit,
        'data_infra': data_infra,
        'overall_score': overall_score
    }

ai_score = calculate_ai_readiness(raw_data['tech_stack'])
print(f'[+] Calculated AI Readiness Score: {ai_score["overall_score"]}/100')
```
